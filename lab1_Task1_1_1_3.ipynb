```python
#3D plot
# Limit x and y from -5 to 5
x = np.linspace(-5, 5, 100)  # 100 points for smooth plot
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)     # create a 2D grid of X and Y
Z = X**2 + Y**2

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

# Plot surface
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')  # smooth colored surface

# Add labels and title
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D plot of Z = X^2 + Y^2')

# Add color bar
fig.colorbar(surf, shrink=0.5, aspect=5)

plt.show()
```

```python
#Random 100000x1 array from the same exponential distribution

# Generate distributions
exp_data = np.random.exponential(scale=1.0, size=100000)   # exponential
uni_data = np.random.rand(100000)                          # uniform [0, 1)
norm_data = np.random.normal(loc=0, scale=1, size=100000)  # normal (mean=0, std=1)

# Plot histograms
plt.hist(exp_data, density=True, bins=100, histtype="step", color="red", label="Exponential")
plt.hist(uni_data, density=True, bins=100, histtype="step", color="blue", label="Uniform")
plt.hist(norm_data, density=True, bins=100, histtype="step", color="green", label="Normal")

# Adjust axis so exponential distribution is visible well
plt.axis([0, 8, 0, 1])  # Only positive values for exponential
plt.legend(loc="upper right")
plt.title("Random distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```
